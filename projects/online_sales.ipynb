```python
import pandas as pd

df = pd.read_csv('online_sales_dataset.csv')
```

```python
#Inspection Data
print('\n------------Info--------------')
print(df.info())
print('\n------------Sample Data--------------')
display(df.sample(5))
```

```python
#  Check Duplicated Data
print(df.duplicated().sum())
print(df[df.duplicated()]  )
```

```python
# Change Data Types
df['InvoiceDate'] = pd.to_datetime(df['InvoiceDate'], errors='coerce')
df.dtypes
```

```python
# Manage String
df.sample(10)
df['PaymentMethod'] = df['PaymentMethod'].str.replace('paypall','Paypal')
```

```python
# Check Null Data
print(df.isnull().sum())
print(df.describe())
```

```python
#Fill Null Value

df['CustomerID'] = df['CustomerID'].fillna('Unknown')
df['ShippingCost'] = df['ShippingCost'].fillna(df['ShippingCost'].median()) 
df['WarehouseLocation'] =df['WarehouseLocation'].fillna('Unknown')

print(df.isnull().sum())
```

```python
# Chaneg Negative Value
df['Quantity'] = df['Quantity'].abs()
df['UnitPrice'] = df['UnitPrice'].abs()
df.describe()
```

```python
# Inspecdata Before Export
df.info()
df.sample(5)
```

```python
# Export to CSV
df.to_csv('cleaned_online_sales.csv', index=False)
```
