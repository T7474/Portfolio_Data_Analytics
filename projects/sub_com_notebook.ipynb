```python
import pandas as pd

df = pd.read_csv('sub_components2025_2026.csv')
```

```python
#Check data frame
df.info()
df.head()
```

```python
# Drop unnecessary columns
df = df.drop(columns=['DEPT','DRAWING','PURLT','SAFETY_STOCK','ITEM_TYPE'])
df.head()
```

```python
# Change Data Type
date_cols = ['DATE', 'LAST_ISSUED_DATE', 'LAST_STOCK_IN_DATE']

for col in date_cols:
    df[col] = pd.to_datetime(df[col], format='%d/%m/%Y', errors='coerce')

qty_amount_cols = ['BEGIN_QTY','RECEIVE_QTY','ISSUE_QTY','STOCK_QTY',
                    'BEGIN_AMOUNT','RECEIVE_AMOUNT','STOCK_AMOUNT','ISSUE_AMOUNT']

df[qty_amount_cols] = df[qty_amount_cols].astype('int64')
df.dtypes
```
