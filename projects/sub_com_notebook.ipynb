```python
import pandas as pd

df = pd.read_csv('sub_components2025_2026.csv')
```

```python
#Check data frame
df.info()
print(df.describe())
df.head()
```

```python
#Check Duplicated Value
print('--------------Check Duplicated Values----------------')
print(df.duplicated().sum())
```

```python
# Drop unnecessary columns
df = df.drop(columns=['DEPT','DRAWING','PURLT','SAFETY_STOCK','ITEM_TYPE'])
df.head()
```

```python
# Change Data Type
date_cols = ['DATE', 'LAST_ISSUED_DATE', 'LAST_STOCK_IN_DATE']

for col in date_cols:
    df[col] = pd.to_datetime(df[col], format='%d/%m/%Y', errors='coerce')

qty_amount_cols = ['BEGIN_QTY','RECEIVE_QTY','ISSUE_QTY','STOCK_QTY']

df[qty_amount_cols] = df[qty_amount_cols].astype('int64')
df.dtypes
```

```python
#Check Null Values
print('--------------Check Null Values----------------')
print(df.isnull().sum())
# Fill null values
df['WC_CODE'] = df['WC_CODE'].fillna('Unknown')
df['UP'] = df['UP'].fillna(df['UP'].median())
df['LAST_ISSUED_DATE'] = df['LAST_ISSUED_DATE'].fillna(df['LAST_ISSUED_DATE'].mode()[0])
print('--------------Null Values After Filling----------------')
print(df.isnull().sum())
```

```python
# Check data frame after cleaning
print('--------------Data Frame After Cleaning----------------')
print(df.info())
display(df.sample(10))
```

```python
#Insert Columns
df['STOCK_AGE'] = (pd.Timestamp.now() - df['LAST_STOCK_IN_DATE']).dt.days // 30
df['GROUP_AGE'] = pd.cut(df['STOCK_AGE'], bins=[0, 3, 6, 12, 24, float('inf')], labels=['0-3 months', '3-6 months', '6-12 months', '12-24 months', '24+ months'])
df['GROUP_AGE'] = df['GROUP_AGE'].astype('category').fillna('0-3 months')

print('--------------Data Frame After Adding Columns----------------')
print(df.info())
display(df[['LAST_STOCK_IN_DATE','STOCK_AGE','GROUP_AGE']].sample(10))
```

```python
df.to_csv('sub_components2025_2026_cleaned.csv', index=False)
```
