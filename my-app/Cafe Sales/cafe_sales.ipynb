```python
import pandas as pd

df = pd.read_csv('dirty_cafe_sales.csv')
```

```python
# Check Data
print("\nDataFrame Info:")
print(df.info())
print("\nUnique Values:")
print(df.nunique())
print("\nFirst 10 Rows:")
display(df.head(10))
```

```python
#Manage Head Column
df.columns = df.columns.str.strip().str.lower().str.replace(' ','_')
print(df.columns)
```

```python
#Check for Duplicates
print(df.duplicated().sum())
display(df[df.duplicated()])
```

```python
#Change Data Type


df['transaction_date'] = pd.to_datetime(df['transaction_date'], format='%Y-%m-%d', errors='coerce')
df['quantity'] = pd.to_numeric(df['quantity'], errors='coerce')
df['price_per_unit'] = pd.to_numeric(df['price_per_unit'], errors='coerce')
df['total_spent'] = pd.to_numeric(df['total_spent'], errors='coerce')

print(df.info())
display(df.head(10))
```

```python
#Check for Missing Values
print(df.isna().sum())

df['item'] = df['item'].fillna('Unknown')
df['quantity'] = df['quantity'].fillna(df['quantity'].median())
df['price_per_unit'] = df.groupby('item')['price_per_unit'].transform(lambda x: x.ffill().bfill())
df['total_spent'] = df['total_spent'].fillna(df['quantity'] * df['price_per_unit'])
df['payment_method'] = df['payment_method'].fillna('Unknown')
df['location'] = df['location'].replace(['ERROR','UNKNOWN'],'Other').fillna('Other')
df['transaction_date'] = df['transaction_date'].fillna('Unknown')

print('\nData After Fill Na')
print(df.isna().sum())
display(df.sample(10))
```

```python
#Manage Unknown Value
import numpy as np

df['item'] = df['item'].replace(['UNKNOWN','ERROR'],'Unknown')
df['payment_method'] = df['payment_method'].replace(['UNKNOWN','ERROR'] , 'Unknown')
df['transaction_date'] = df['transaction_date'].replace('UNKNOWN' , np.nan).ffill()

display(df.sample(10))
```

```python
# Check Dataframe
print('\n---------------- Data Frame Information--------------------')
print(df.info())
print('\n----------------Missing Value--------------------------')
print(df.isna().sum())
print('\n-----------------Unique Value------------------------')
print(df.nunique())
print('\n-----------------Sample Value------------------------')
display(df.sample(5))
```

```python
df['transaction_date'] = pd.to_datetime(df['transaction_date'], errors='coerce')
df['transaction_date'] = df['transaction_date'].fillna(df['transaction_date'].ffill())
print(df.info())
```

```python
# ดูจำนวนแถวทั้งหมดที่เป็น 'ERROR'
print((df['item'] == 'ERROR').sum())
print((df['payment_method'] == 'ERROR').sum())

# หรือกรณีที่มีคำว่า 'ERROR' ปนอยู่กับข้อความอื่น (เช่น 'ERROR_404', 'SYSTEM ERROR')
error_rows = df[df['item'].astype(str).str.contains('ERROR', case=False, na=False)]
```

```python
df.to_csv('cleaned_cafe_sales.csv', index=False)
```
