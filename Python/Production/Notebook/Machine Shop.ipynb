```python
import pandas as pd
import numpy as np

df = pd.read_csv("Monthly_Production.csv")
df.info()
df.sample(10)
```

```python
#ลบคอลัมน์ที่มีแถวว่างทั้งหมด

df.isnull().sum()

df = df.dropna(axis=1, how="all")
```

```python
#เปลี่ยนชนิดข้อมูลเป็น string

df['AC_PROCESS'] = df['AC_PROCESS'].astype('str')

print(df.isnull().sum())
df.info()
display(df.sample(5))
```

```python
#เติมช่องว่างด้วย Unknown

str_columns = df.select_dtypes(include=['object']).columns

df[str_columns] = df[str_columns].fillna('Unknown')

print(df.isna().sum())
print(df.info())
df.sample(10)
```

```python
#เช็คคอลัมน์ที่มีตัวเลขติดลบ

number_df = df.select_dtypes(include=['number'])
column_negative = number_df.columns[(number_df < 0).any()].tolist()



print("คอลัมน์ที่มีข้อมูลติดลบ:", column_negative)
print((df['NG_QTY']<0).sum())
display(df.sample(10))
```

```python
#แปลงข้อมูลติดลบเป็นบวก

df['NG_QTY'] = df['NG_QTY'].abs()
print((df['NG_QTY']<0).sum())
display(df.sample(10))
```

```python
#ตรวจสอบข้อมูลก่อนส่งออก

print(df.info())
print(df.isnull().sum())
print(f"จำนวนแถวคงเหลือ: {len(df)}")
```

```python
df.to_csv("Monthly_Production_Clean.csv", index=False)
```
