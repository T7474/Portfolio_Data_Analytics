```python
import pandas as pd

df = pd.read_csv('../Data/PO_Outgoing_List.csv')

print("\n ----------รายละเอียด----------")
print(df.info())
print("\n ----------จำนวนแถวและคอมลัมน์----------")
print(df.shape)
print("\n ----------จำนวนข้อมูลซ้ำ----------")
print(df.duplicated().sum())
print("\n ----------จำนวนข้อมูลว่าง----------")
print(df.isna().sum())
print("\n ----------ตัวอย่างข้อมูล----------")
display(df.sample(5))
```

```python
#แปลงชนิดข้อมูลคอมลัมน์ SECTION และ VENDOR เป็น string

df = df.astype({'SECTION':'str','VENDOR':'str'})
print("\n -----------รายละเอียด-----------")
print(df.info())
```

```python
#แปลงชนิดข้อมูลวันที่
df['SENT_DATE'] = pd.to_datetime(df['SENT_DATE'],errors='coerce')
df['DUEDATE'] = pd.to_datetime(df['DUEDATE'],errors='coerce')
df['SENDING_DATETIME'] = df['SENDING_DATETIME'].astype('str').str[:10]
df['SENDING_DATETIME'] = pd.to_datetime(df['SENDING_DATETIME'],errors='coerce')

print("\n ตรวจสอบข้อมูล")
print(df.info())
print('\n ตัวอย่างข้อมูล')
display(df.sample(5))
```

```python
#ลบคอลัมน์ที่ไม่มีข้อิมูลเลย เติมข้อมูลในแถวว่าง
df = df.dropna(axis=1, how='all')
df = df.fillna({'ITEM_NO':'Unknown', 'SPEC':'Unknown'})

print('\nจำนวนข้อมูลว่าง')
print(df.isna().sum())
display(df.sample(5))
```

```python
# นับจำนวนข้อมูลที่ติดลบแยกตามรายคอลัมน์ และตรวจสอบผลลัพธ์
print((df[['UNIT_PRICE', 'ORDER_QTY', 'ORDER_AMT']] < 0).sum())
df.info()
```

```python
# ลบแถวที่ PONO ขึ้นต้นด้วย STA หรือ DPB (เลือกเก็บเฉพาะแถวที่ไม่ใช่)
df = df[~df['PONO'].str.startswith(('STA'))]
```

```python
#เพิ่มคอลัมน์แผนก

Department_Map = {'AA' : ' Assembly', 
                  'FA' : 'Asembly', 
                  'RG' : 'Rouge Grinding', 
                  'CF' : 'Cold Forming',
                  'TN' : 'Turning',
                  'HT' : 'Heat Treatment',
                  'BR' : 'Heat Treatment',
                  'MM' : 'Maintenance 41',
                  'FM' : 'Maintenance 41',
                  'MS' : 'Maintenance 40',
                  'MF' : 'Maintenance 40',
                  'QC' : 'Quility Control',
                  'QM' : 'Quility Control',
                  'EN' : 'Engineer',
                  'PC' : 'Production Control',
                  'QG' : 'Gauge Control',
                  'SG' : '2nd Grinding',
                  'FG' : '1st Grinding',
                  'FO' : '1st Grinding',
                  'FF' : '1st Grinding',
                  'TC' : 'Tooling Center',
                  'SS' : 'Store',
                  'PP' : 'PP',
                  'DP' : 'Despatch'}

df['Department'] = df['PONO'].str[:2].map(Department_Map)

df.info()
df.sample(5)
```

```python
#เติมช่องว่าง

df = df.dropna(subset=['Department'])
print(df.isnull().sum())



```

```python
#ส่งออกไฟล์
df.to_csv('../Data/PO_Outgoing_List_Clean.csv', index=False)
```
