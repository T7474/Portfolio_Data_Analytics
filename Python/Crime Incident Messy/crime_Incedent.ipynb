```python
#Import Pandas Library
import pandas as pd

#Import CSV file
df = pd.read_csv('crime_incidents.csv')

print('\n จำนวนแถวและคอลัมน์ของ DataFrame:')
print(df.shape)
print('\n ชนิดข้อมูลและจำนวนค่าที่ไม่ใช่ค่า null ของแต่ละคอลัมน์:')
print(df.info())
print('\n ข้อมูลที่ซ้ำกัน:')
print(df.duplicated().sum())
print('\n ตัวอย่างข้อมูล 5 แถวแรกของ DataFrame:')
display(df.sample(5))
```

```python
#Check duplicated rows
print(df.duplicated().sum())
print(df.shape)

#Delete duplicated rows
df = df.drop_duplicates()
print(df.shape)
```

```python
#Change data types
df['incident_datetime'] = pd.to_datetime(df['incident_datetime'],errors='coerce')
df['property_loss_usd'] = pd.to_numeric(df['property_loss_usd'], errors='coerce')
df['victim_age'] = pd.to_numeric(df['victim_age'], errors='coerce')

df.info()
```

```python
# Check and Fill missing values

df = df.fillna({
    'latitude': df['latitude'].mean(),
    'longitude': df['longitude'].mean(),
    'incident_datetime' : df['incident_datetime'].ffill(),
    'badge_number' : df['badge_number'].mean(),
    'suspect_id' : 'unknown',
    'suspect_first_name' : 'unknown',
    'suspect_last_name' : 'unknown',
    'suspect_age' : df['suspect_age'].mean(),
    'suspect_gender' : 'unknown',
    'suspect_race' : 'unknown',
    'victim_id' : 'unknown',
    'victim_first_name' : 'unknown',
    'victim_last_name' : 'unknown',
    'victim_age' : df['victim_age'].mean(),
    'victim_gender' : 'unknown',
    'victim_phone' : 'unknown',
    'weapon_used' : 'unknown',
    'severity' : 'unknown',
    'case_status' : 'unknown',
    'resolution' : 'unknown',
    'num_arrests' : 0,
    'property_loss_usd' : df['property_loss_usd'].mean(),
    'reported_online' : 'unknown',
    'notes' : 'unknown'
})

print('\n-------------ข้อมูลหลังการเติมข้อมูล-------------')
print(df.isna().sum())
```

```python
# Change to lower string
pd.set_option('display.max_columns', 35)


column_str = ['crime_type', 'district', 'city', 'suspect_gender', 'suspect_race', 'victim_gender', 'severity', 'resolution','reported_online','case_status','weapon_used']

# ใช้ .apply เพื่อสั่งให้ทุกคอลัมน์ในรายชื่อแปลงเป็นตัวพิมพ์เล็ก
df[column_str] = df[column_str].apply(lambda x: x.str.lower())
```

```python
df.sample(10)
```

```python
#Change word
# ใช้ .replace() ตรงๆ โดยไม่ต้องใส่ .str และเพิ่ม regex=False เพื่อความแม่นยำ
df['district'] = df['district'].str.strip()
df['district'] = df['district'].replace({'nor':'north','sou':'south', 'cen' : 'central','eas':'east', 'mid' : 'midtown', 'wes' : 'west'})
df['suspect_gender'] = df['suspect_gender'].replace({'m':'male', 'f':'female'}, regex=False)
df['severity'] = df['severity'].replace({'med':'medium','crit':'critical','1':'low', '2':'medium','3':'high', '4':'critical'}, regex=False)
df['case_status'] = df['case_status'].replace({'pendng':'pending','under investigation' : 'investigation','investgation':'investigation'}, regex=False)
df['reported_online'] = df['reported_online'].replace({'true':'yes','false':'no', '1':'yes', '0':'no'}, regex=False)
df['resolution'] = df['resolution'].replace({'case dismissed' : 'dismissed', 'arres made' : 'arrested','arrest made':'arrested', 'warning' : 'warning issued'}, regex=False)
df['victim_gender'] = df['victim_gender'].replace({'m':'male', 'f':'female'}, regex=False)
```

```python
df.sample(10)
```

```python
# Change negative values to positive
df['victim_age'] = df['victim_age'].abs()
df['suspect_age'] = df['suspect_age'].abs()
df['property_loss_usd'] = df['property_loss_usd'].abs()
df['num_arrests'] = df['num_arrests'].abs()

print((df['victim_age']<0).sum())
print((df['suspect_age']<0).sum())
print(df.info())
```

```python
# รันโค้ดนี้หลังจากที่ทำการโหลด df ใหม่เรียบร้อยแล้ว
df['victim_age_group'] = pd.cut(df['victim_age'], bins=[0, 12, 17, 24, 34, 49, 64, float('inf')], labels=['0-12', '13-17', '18-24', '25-34', '35-49', '50-64', '65+'])
df['suspect_age_group'] = pd.cut(df['suspect_age'], bins=[0, 12, 17, 24, 34, 49, 64, float('inf')], labels=['0-12', '13-17', '18-24', '25-34', '35-49', '50-64', '65+'])

# ทดลองแสดงผลตรวจสอบ 5 แถวแรก
display(df[['victim_age', 'victim_age_group', 'suspect_age', 'suspect_age_group']].head())
```

```python
#IQR
Q1 = df['victim_age'].quantile(0.25)
Q3 = df['victim_age'].quantile(0.75)
IQR = Q3 - Q1
lower = Q1 - 1.5 * IQR
upper = Q3 + 1.5 * IQR
df = df[(df['victim_age'] >= lower) & (df['victim_age'] <= upper)]
df = df[(df['suspect_age'] >= lower) & (df['suspect_age'] <= upper)]
```

```python
print((df['victim_age'] > 100).sum(),(df['suspect_age'] > 100).sum())
display(df[df['victim_age'] > 100]['victim_age'])
display(df[['victim_age','suspect_age']].describe())
print(df.info())
```

```python
#Export to CSV file
df.to_csv('cleaned_crime_incidents.csv', index=False)
```
