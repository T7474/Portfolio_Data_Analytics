```python
import pandas as pd

df = pd.read_csv('crime_incidents.csv')

print('\n จำนวนแถวและคอลัมน์ของ DataFrame:')
print(df.shape)
print('\n ชนิดข้อมูลและจำนวนค่าที่ไม่ใช่ค่า null ของแต่ละคอลัมน์:')
print(df.info())
print('\n ข้อมูลที่ซ้ำกัน:')
print(df.duplicated().sum())
print('\n ตัวอย่างข้อมูล 5 แถวแรกของ DataFrame:')
display(df.sample(5))
```

```python
#ลบแถวซ้ำ
df.duplicated().sum()
print(df.shape)

df = df.drop_duplicates()
print('\n จำนวนแถวและคอลัมน์ของ DataFrame หลังจากลบแถวซ้ำ:')
print(df.shape)
```

```python
#เปลี่ยนชนิดข้อมูล


df['incident_datetime'] = pd.to_datetime(df['incident_datetime'],errors='coerce')
df['property_loss_usd'] = pd.to_numeric(df['property_loss_usd'], errors='coerce')

df.info()
```

```python
# เติมข้อมูลหาย


num_cols = df.select_dtypes(include=['number']).columns
df[num_cols] = df[num_cols].fillna(df[num_cols].mean())

str_cols = df.select_dtypes(include=['object']).columns
df[str_cols] = df[str_cols].fillna('Unknown')
print(df.isna().sum())

df['incident_datetime'] = df['incident_datetime'].ffill()
```
